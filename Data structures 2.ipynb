## Dictionary

```python
student_marks={'Anu':80,'Kavya':75,'Anjali':70,'Rahul':65,'Tina':60}
student_marks
```

```python
student_marks['Kavya']
```

```python
student_marks['Janani']=80
student_marks
```

```python
print('Students name:',student_marks.keys())
print('Students mark:',student_marks.values())
print('Students name and marks:',student_marks.items())
```

## Sets

```python
my_set={'a','e','i','o','u','a','a','i'}
my_set 
''' a set does not allow duplicates i.e each item in a set is unique.
The items 'a'and'i' appeared multiple times in the set,but the output doesn't allow it and it appear only one time.'''
```

```python
my_set[4]='s'
my_set
''' sets are unordered collection and does not support indexing.
so we can't modify the set using index position.'''
```

```python
set1={1,3,5,7,9}
set2={2,3,5,8,10}
print(set1)
print(set2)
```

```python
union=set1.union(set2)
print('union:',set1.union(set2))
```

```python
print('intersection:',set1.intersection(set2))
```

## Operators and conditional statements

```python
Score=int(input('Enter your score (0to 10)'))

if Score < 0 or Score > 10:
    print('Invalid Score! : Enter Score between 0 and 10.')
elif Score > 7:
    print('Above Average : Excellent and keep it up.')
elif Score >= 4:
    print('Average : Good effort and keep practicing.')
elif Score < 4:
    print('Below Average : Need to improve your performance,consistent practice will lead to better results.')

```
